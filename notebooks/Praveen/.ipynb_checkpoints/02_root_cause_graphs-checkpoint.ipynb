```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

# ==========================================
# STEP 1: Load One Experiment Folder
# ==========================================
# Update this path to wherever your specific folder is located
FOLDER_PATH = '../../data/raw/tracebench/AN_Data_corruptBlk_r_00FDN_30C_1to19B_0to120INT_15RT_5WT/'

df_events = pd.read_csv(FOLDER_PATH + 'event.csv')
df_edges = pd.read_csv(FOLDER_PATH + 'edge.csv')
```

```python
# ==========================================
# STEP 2: Isolate a Single Trace (TaskID)
# ==========================================
# We only want to map one specific failure timeline to keep the graph readable
TARGET_TASK = 'B076E6516B275ABB'

events_filtered = df_events[df_events['TaskID'] == TARGET_TASK]
edges_filtered = df_edges[df_edges['TaskID'] == TARGET_TASK]

print(f"Isolated Task: {TARGET_TASK}")
print(f"Nodes found: {len(events_filtered)} | Edges found: {len(edges_filtered)}")
```

```python
# ==========================================
# STEP 3: Build the Directed Graph
# ==========================================
G = nx.DiGraph()

# Add Nodes (Using TID as the unique ID, and OpName as the human-readable label)
for index, row in events_filtered.iterrows():
    G.add_node(row['TID'], label=row['OpName'])

# Add Edges (Drawing arrows from Father to Child)
for index, row in edges_filtered.iterrows():
    G.add_edge(row['FatherTID'], row['ChildTID'])
```

```python
# ==========================================
# STEP 4: Visualize the Topology
# ==========================================
plt.figure(figsize=(12, 8))

# Extract the labels we assigned earlier so they show up on the graph
node_labels = nx.get_node_attributes(G, 'label')

# Calculate the layout (spring_layout pushes nodes apart visually)
pos = nx.spring_layout(G, k=0.5, iterations=50)

# Draw the Graph!
nx.draw(G, pos, 
        labels=node_labels, 
        with_labels=True, 
        node_color='lightcoral', 
        edge_color='gray', 
        node_size=3000, 
        font_size=8, 
        font_weight='bold',
        arrows=True,
        arrowsize=20)

plt.title(f"Root Cause Analysis Causal Graph for Task: {TARGET_TASK}")
plt.show()
```
